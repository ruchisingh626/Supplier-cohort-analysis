```python
category_sales = df.groupby('category')['units_sold'].sum().sort_values(ascending=False)

print(category_sales)
```

```python
plt.figure(figsize=(9,5))

category_sales.plot(kind='bar')

plt.title('Total Units Sold by Category')
plt.xlabel('Product Category')
plt.ylabel('Total Units Sold')
plt.xticks(rotation=45)

plt.show()
```

```python
inventory_category = df.groupby('category')['inventory_level'].mean().sort_values(ascending=False)

print(inventory_category)
```

```python
plt.figure(figsize=(9,5))

inventory_category.plot(kind='bar')

plt.title('Average Inventory Level by Category')
plt.xlabel('Category')
plt.ylabel('Average Inventory')

plt.show()
```

```python
supplier_analysis = df.groupby('supplier').agg({
    'units_sold': 'sum',
    'lead_time_days': 'mean',
    'transportation_cost': 'mean',
    'order_accuracy': 'mean'
})

print(supplier_analysis)
```

```python
transport_analysis = df.groupby('region')['transportation_cost'].sum().sort_values(ascending=False)

print(transport_analysis)
```

```python
plt.figure(figsize=(9,5))

transport_analysis.plot(kind='bar')

plt.title('Transportation Cost by Region')
plt.xlabel('Region')
plt.ylabel('Total Transportation Cost')

plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

correlation = numeric_df.corr()

print(correlation)
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Inventory & Supply Chain Correlation Heatmap')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.boxplot(
    data=df,
    x='backorder',
    y='lead_time_days'
)

plt.title('Lead Time vs Backorder')
plt.xlabel('Backorder')
plt.ylabel('Lead Time (Days)')

plt.show()
```

```python
plt.figure(figsize=(9,5))

sns.scatterplot(
    data=df,
    x='inventory_level',
    y='units_sold',
    hue='category'
)

plt.title('Inventory Level vs Units Sold')
plt.xlabel('Inventory Level')
plt.ylabel('Units Sold')

plt.show()
```

```python
backorder_insights = df.groupby('category').agg(
    total_orders=('category', 'count'),
    backorders=('backorder', 'sum')
)

backorder_insights['backorder_rate'] = (
    backorder_insights['backorders'] /
    backorder_insights['total_orders']
) * 100

backorder_insights = backorder_insights.sort_values(
    'backorder_rate', ascending=False
)

print(backorder_insights)
```

```python
supplier_insights = df.groupby('supplier').agg(
    avg_lead_time=('lead_time_days', 'mean'),
    order_accuracy=('order_accuracy', 'mean'),
    backorder_rate=('backorder', 'mean')
)

supplier_insights['order_accuracy'] *= 100
supplier_insights['backorder_rate'] *= 100

print(supplier_insights.round(2))
```

```python
warehouse_insights = df.groupby('warehouse').agg(
    avg_inventory=('inventory_level', 'mean'),
    avg_capacity=('warehouse_capacity', 'mean')
)

warehouse_insights['utilization_rate'] = (
    warehouse_insights['avg_inventory'] /
    warehouse_insights['avg_capacity']
) * 100

print(warehouse_insights.round(2))
```

```python
accuracy = df['order_accuracy'].value_counts()

print(accuracy)
```

```python
accuracy_percentage = df['order_accuracy'].mean() * 100

print("Overall Order Accuracy:", round(accuracy_percentage, 2), "%")
```

```python
kpis = {
    'Total Units Sold': df['units_sold'].sum(),
    'Average Inventory': df['inventory_level'].mean(),
    'Total Transportation Cost': df['transportation_cost'].sum(),
    'Average Lead Time': df['lead_time_days'].mean(),
    'Backorder Rate': df['backorder'].mean() * 100,
    'Order Accuracy': df['order_accuracy'].mean() * 100
}

for key, value in kpis.items():
    print(f"{key}: {value:.2f}")
```

```python
print("========== SUPPLY CHAIN KPI REPORT ==========")

print("Total Units Sold:", df['units_sold'].sum())

print("Average Inventory:",
      round(df['inventory_level'].mean(), 2))

print("Average Lead Time:",
      round(df['lead_time_days'].mean(), 2))

print("Total Transportation Cost:",
      round(df['transportation_cost'].sum(), 2))

print("Backorder Rate:",
      round(df['backorder'].mean() * 100, 2), "%")

print("Order Accuracy:",
      round(df['order_accuracy'].mean() * 100, 2), "%")
```

```python
region_sales = df.groupby('region')['units_sold'].sum().sort_values(ascending=False)

plt.figure(figsize=(9,5))

sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Units Sold')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.barplot(
    data=df,
    x='supplier',
    y='order_accuracy',
    errorbar=None
)

plt.title('Supplier Order Accuracy Comparison')
plt.xlabel('Supplier')
plt.ylabel('Order Accuracy')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(7,5))

df['backorder'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    labels=['No Backorder', 'Backorder']
)

plt.title('Backorder Distribution')
plt.ylabel('')

plt.show()
```

```python
df['month'] = df['date'].dt.to_period('M')

monthly_sales = df.groupby('month')['units_sold'].sum()

plt.figure(figsize=(14,5))

monthly_sales.plot(kind='line', marker='o')

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Units Sold')

plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.show()
```
